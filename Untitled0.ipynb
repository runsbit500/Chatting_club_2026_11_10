```python
# 必要なライブラリのインストール（音声合成・自動翻訳）
!pip install -q gTTS deep-translator

import base64
import io
import json
import time
from gtts import gTTS
from deep_translator import GoogleTranslator
from google.colab import files

# ==========================================
# ★ 毎回ここだけ書き換えてください ★
# ==========================================
LESSON_DATE = "2026/11/18"

ENGLISH_SENTENCES = [
    "What I want is your love.",
    "It's not I but you that are wrong.",
    "I've never seen a bird like that.",
    "There is nothing like a beer after work.",
    "As I told you before, I am not a manager here."
]
# ==========================================

translator = GoogleTranslator(source='en', target='ja')
lessons_data = []

print("翻訳と音声データの生成を開始します...")
# 実行直後の最初のアクション前に、安全のため少し待機
time.sleep(3.0)

for i, eng in enumerate(ENGLISH_SENTENCES, 1):
    eng_clean = eng.strip()
    if not eng_clean:
        continue

    # 1. 日本語に自動翻訳（極めて強力なアクセス制限回避ロジック）
    jpn = ""
    try:
        jpn = translator.translate(eng_clean)
    except Exception as e:
        # 1度目のエラー時：一過性のアクセス制限解除を待つため60秒間、大きく待機時間を取ります
        print(f"翻訳エラーが発生しました。アクセス制限解除のため、60秒間しっかり待機して再試行します... (Error: {e})")
        time.sleep(60.0)
        try:
            jpn = translator.translate(eng_clean)
        except Exception as e2:
            # 2度目もダメだった場合：強制終了を回避し、フォールバックテキストを設定して進めます
            print(f"再試行も失敗しました。翻訳なしで処理を続行します... (Error: {e2})")
            jpn = "[自動翻訳エラー: 翻訳が利用できませんでした]"

    # 2. 英語MP3音声を生成してBase64に変換
    tts = gTTS(text=eng_clean, lang='en', tld='us')
    mp3_fp = io.BytesIO()
    tts.write_to_fp(mp3_fp)
    mp3_fp.seek(0)
    b64_str = base64.b64encode(mp3_fp.read()).decode('utf-8')
    audio_data_uri = f"data:audio/mpeg;base64,{b64_str}"

    lessons_data.append({
        "english": eng_clean,
        "japanese": jpn,
        "audio": audio_data_uri
    })
    print(f"[{i}/{len(ENGLISH_SENTENCES)}] 完了: {eng_clean} -> {jpn}")

    # サーバーへの負荷を完全に抑えるため、各文の処理の間に必ず5秒間のウェイトを置きます
    time.sleep(5.0)

lessons_json = json.dumps(lessons_data, ensure_ascii=False)
desc_text = f"{len(lessons_data)}つの英文を、和訳と読み上げで練習できます。"

html_template = f"""<!DOCTYPE html>
<html lang="ja">
<head>
<meta charset="utf-8"/>
<meta content="width=device-width, initial-scale=1" name="viewport"/>
<meta content="{desc_text}" name="description"/>
<meta content="#f2f4e9" name="theme-color"/>
<title>Lesson {LESSON_DATE}</title>
<style>
:root {{ color-scheme: light; --ink: #20342d; --green: #245b47; --muted: #50635a; --line: #d6dfd2; --paper: #fffef9; }}
* {{ box-sizing: border-box; }}
body {{ margin: 0; background: #f2f4e9; color: var(--ink); font: 16px/1.65 system-ui, -apple-system, sans-serif; }}
main {{ max-width: 44rem; margin: auto; padding: clamp(18px, 5vw, 42px); }}
header {{ display: flex; align-items: baseline; justify-content: space-between; flex-wrap: wrap; gap: 4px 12px; margin-bottom: 14px; }}
h1 {{ margin: 0; font-size: clamp(1.4rem, 5vw, 2rem); }}
#counter {{ color: var(--muted); font-weight: 750; white-space: nowrap; }}
article {{ background: var(--paper); border: 1px solid var(--line); border-radius: 22px; padding: clamp(18px, 5vw, 34px); box-shadow: 0 7px 18px #20342d08; }}
.english {{ margin: 0; font: 600 clamp(1.7rem, 5.4vw, 2.6rem)/1.26 Georgia, serif; overflow-wrap: anywhere; }}
.japanese {{ color: var(--muted); margin: 14px 0 0; }}
.actions {{ display: flex; align-items: center; flex-wrap: wrap; gap: 8px 16px; margin-top: 20px; }}
.japanese-toggle {{ display: inline-flex; align-items: center; gap: 8px; min-height: 48px; padding: 0 6px; cursor: pointer; font-weight: 700; }}
.japanese-toggle input {{ width: 20px; height: 20px; margin: 0; accent-color: var(--green); cursor: pointer; }}
.japanese-toggle input:focus-visible {{ outline: 3px solid #987018; outline-offset: 3px; }}
button {{ font: inherit; font-weight: 700; color: var(--ink); background: var(--paper); border: 1px solid var(--line); border-radius: 12px; min-height: 48px; padding: 8px 15px; cursor: pointer; touch-action: manipulation; }}
button:hover:not(:disabled) {{ border-color: var(--green); background: #e9f0e6; }}
button:focus-visible {{ outline: 3px solid #987018; outline-offset: 3px; }}
button:disabled {{ opacity: .5; cursor: default; }}
#listen {{ background: var(--green); border-color: var(--green); color: white; }}
#listen:hover {{ background: #174b39; }}
#listen svg {{ width: 18px; height: 18px; vertical-align: -3px; margin-right: 5px; }}
#speechStatus {{ margin: 10px 0 0; color: #783d30; font-size: .9rem; }}
nav {{ display: flex; gap: 8px; margin: 16px 0; }}
nav button {{ flex: 1; min-width: 0; }}
</style>
</head>
<body>
<main>
<header>
<h1 lang="en">Lesson {LESSON_DATE}</h1>
<span aria-label="文の番号" id="counter"></span>
</header>
<nav aria-label="上部の文切り替え">
<button id="prevTop" type="button">← 前へ</button>
<button id="nextTop" type="button">次へ →</button>
</nav>
<article aria-label="練習する文">
<div aria-atomic="true" aria-live="polite">
<p class="english" id="english" lang="en-US"></p>
<p class="japanese" hidden="" id="japanese"></p>
</div>
<div class="actions">
<label class="japanese-toggle">
<input aria-controls="japanese" autocomplete="off" id="showJapanese" type="checkbox"/>
        日本語
      </label>
<button id="listen" type="button"><svg aria-hidden="true" fill="none" stroke="currentColor" stroke-linecap="round" stroke-linejoin="round" stroke-width="2" viewbox="0 0 24 24"><path d="M3 9h4l5-4v14l-5-4H3zM16 9a4 4 0 0 1 0 6m3-9a8 8 0 0 1 0 12"></path></svg>読み上げ</button>
</div>
<p hidden="" id="speechStatus" role="status"></p>
</article>
<nav aria-label="練習する文を切り替える">
<button id="prev" type="button">← 前へ</button>
<button id="next" type="button">次へ →</button>
</nav>
</main>
<script>
const lessons = {lessons_json};

const el = id => document.getElementById(id);
const player = new Audio();
let index = 0;

function stopAudio() {{
  player.pause();
  player.currentTime = 0;
  if ('speechSynthesis' in window) {{
    window.speechSynthesis.cancel();
  }}
}}

function renderLesson() {{
  const lesson = lessons[index];
  stopAudio();
  el("speechStatus").hidden = true;
  el("counter").textContent = `${{index + 1}} / ${{lessons.length}} 文`;
  el("english").textContent = lesson.english;
  el("japanese").textContent = lesson.japanese;
  el("prev").disabled = index === 0;
  el("next").textContent = index === lessons.length - 1 ? "最初に戻る" : "次へ →";
}}

el("showJapanese").addEventListener("change", event => {{
  el("japanese").hidden = !event.currentTarget.checked;
}});

el("listen").addEventListener("click", () => {{
  const status = el("speechStatus");
  status.hidden = true;
  const lesson = lessons[index];
  stopAudio();
  if (lesson.audio) {{
    player.src = lesson.audio;
    player.play().catch(() => {{
      status.textContent = "音声を再生できませんでした。端末の音量設定をご確認ください。";
      status.hidden = false;
    }});
  }} else if ('speechSynthesis' in window) {{
    const utter = new SpeechSynthesisUtterance(lesson.english);
    utter.lang = 'en-US';
    utter.rate = 0.95;
    window.speechSynthesis.speak(utter);
  }} else {{
    status.textContent = "このブラウザは音声読み上げに対応していません。";
    status.hidden = false;
  }}
}});

el("prev").addEventListener("click", () => {{
  index--;
  renderLesson();
  if (index === 0) el("next").focus();
}});

el("next").addEventListener("click", () => {{
  index = (index + 1) % lessons.length;
  renderLesson();
}});

document.addEventListener("keydown", (e) => {{
  if (e.key === "ArrowLeft" && index > 0) el("prev").click();
  if (e.key === "ArrowRight") el("next").click();
}});

renderLesson();

const prevTop = el('prevTop'), nextTop = el('nextTop');
prevTop.addEventListener('click', () => el('prev').click());
nextTop.addEventListener('click', () => el('next').click());
function syncTopButtons() {{
  prevTop.disabled = el('prev').disabled;
  nextTop.textContent = el('next').textContent;
}}
new MutationObserver(syncTopButtons).observe(el('prev'), {{ attributes: true, attributeFilter: ['disabled'] }});
new MutationObserver(syncTopButtons).observe(el('next'), {{ childList: true }});
syncTopButtons();
</script>
</body>
</html>
"""

output_filename = "index.html"
with open(output_filename, "w", encoding="utf-8") as f:
    f.write(html_template)

print(f"\n生成完了！ {output_filename} をダウンロードします...")
files.download(output_filename)
```
